#### a. Boxplot

````python
### Chọn các numeric columns (trừ Class and Id)
# Get all numerical columns: 
numerical_cols = data_train.select_dtypes(include='number').columns.tolist()
# Drop columns: Class & Id
removed_cols = ['Class', 'Id']
numerical_cols = [col for col in numerical_cols if col not in removed_cols]
# print(numerical_cols)

# Boxplot for each feature by class:
BOXPLOT_N_COLUMNS = 4
fig, axes = plt.subplots(ncols=BOXPLOT_N_COLUMNS, nrows=len(numerical_cols)//BOXPLOT_N_COLUMNS + 1, figsize=(20, 5*len(numerical_cols)//BOXPLOT_N_COLUMNS))

# axes: many axis, because our figure have a lots subplot, so axes is 2D array.
axes = axes.flatten()   # flatten 2D array -> 1D array

for idx, col in enumerate(numerical_cols):
    sns.boxplot(data=data_train, x='Class', y=col, ax=axes[idx])    # ax: current subplot which we are interacting in our figure
    axes[idx].set_title(f"{col} by Class")
    axes[idx].tick_params(axis='x', rotation=25)    # rotate x axis 25* counterclockwise (left rotation) (clockwise - right rotation)

# Hide abundance subplot
for idx in range(len(numerical_cols), len(axes)):
    axes[idx].set_visible(False)

plt.tight_layout()
plt.savefig(REPORTS_PATH / "boxplots_by_class.png")
plt.show();
````

#### d) Demensionality reduction

````python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Chuẩn bị dữ liệu
X = data_train[numerical_cols].fillna(data_train[numerical_cols].median())
y = data_train['Class']

# Chuẩn hóa dữ liệu (quan trọng cho PCA)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Áp dụng PCA giảm xuống 2 chiều
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Visualize
plt.figure(figsize=(12, 8))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap='tab10', alpha=0.6, edgecolors='k')
plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%} variance)')
plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%} variance)')
plt.title('PCA: Class Separation Visualization')
plt.colorbar(scatter, label='Class')
plt.grid(alpha=0.3)
plt.savefig(REPORTS_PATH / "pca_class_separation.png")
plt.show()

# Xem tỷ lệ variance được giải thích
print(f"Total variance explained: {pca.explained_variance_ratio_.sum():.2%}")
````

````python
from sklearn.manifold import TSNE

# Áp dụng t-SNE
tsne = TSNE(n_components=2, random_state=42, perplexity=30)
X_tsne = tsne.fit_transform(X_scaled)

# Visualize
plt.figure(figsize=(12, 8))
scatter = plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=y, cmap='tab10', alpha=0.6, edgecolors='k')
plt.xlabel('t-SNE Component 1')
plt.ylabel('t-SNE Component 2')
plt.title('t-SNE: Class Separation Visualization')
plt.colorbar(scatter, label='Class')
plt.grid(alpha=0.3)
plt.savefig(REPORTS_PATH / "tsne_class_separation.png")
plt.show()
````

````python
# Cần cài đặt: pip install umap-learn
import umap

# Áp dụng UMAP
reducer = umap.UMAP(n_components=2, random_state=42, n_neighbors=15, min_dist=0.1)
X_umap = reducer.fit_transform(X_scaled)

# Visualize
plt.figure(figsize=(12, 8))
scatter = plt.scatter(X_umap[:, 0], X_umap[:, 1], c=y, cmap='tab10', alpha=0.6, edgecolors='k')
plt.xlabel('UMAP Component 1')
plt.ylabel('UMAP Component 2')
plt.title('UMAP: Class Separation Visualization')
plt.colorbar(scatter, label='Class')
plt.grid(alpha=0.3)
plt.savefig(REPORTS_PATH / "umap_class_separation.png")
plt.show()
````

- Comment: 
    + Via PCA, t-SNE, UMAP, we see some clusters but also significant overlap between classes
    + The relationship between features and classes is non-linear.

## 2.2 Handle missing values

### 2.2.1 Identify missing values

````python
def missing_report(df):
    """
    Check and report missing value ins DF
    - Input: df - Pandas DataFrame
    - Output: missing_report - pd with 2 cols: 'Count' & 'Percentage'
    """
    missing_count = df.isna().sum()
    missing_percentage = (missing_count / len(df)) * 100

    missing_report = pd.concat([missing_count, missing_percentage], axis=1)
    missing_report.columns = ['Count', 'Percentage']

    return missing_report

missing_report(data_train)
````
