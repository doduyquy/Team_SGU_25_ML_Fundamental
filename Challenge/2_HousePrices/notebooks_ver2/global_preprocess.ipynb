# 🔧 Demo Preprocessing Modules - House Prices Dataset

Notebook này demo các chức năng của 3 modules preprocessing:
- 🧹 **Clean**: Xử lý dữ liệu thiếu, duplicate, balance
- 🔄 **Preprocess**: Scaling, encoding, outlier handling
- ⭐ **Feature**: Feature selection và engineering

```python
# Import thư viện cần thiết
import pandas as pd
import numpy as np
import sys
import os

# Thêm đường dẫn src vào sys.path
sys.path.append(os.path.abspath('../src'))

# Import các module preprocessing
from preprocess.clean.clean import Clean
from preprocess.preprocess.preprocess import Preprocess
from preprocess.feature.feature import Feature

print("✅ Import thành công!")
print("📦 Modules: Clean, Preprocess, Feature")
```

## 📂 Load Dữ Liệu

Load dữ liệu train và test từ thư mục data/raw/

```python
# Load dữ liệu train và test
train_path = '../data/raw/train.csv'
test_path = '../data/raw/test.csv'

df_train = pd.read_csv(train_path)
df_test = pd.read_csv(test_path)

print(f"📊 Train shape: {df_train.shape}")
print(f"📊 Test shape: {df_test.shape}")

# Hiển thị một vài dòng đầu
print("\n🔍 Xem dữ liệu train:")
df_train.head()
```

# 🧹 PHẦN 1: CLEAN MODULE

Module Clean xử lý các vấn đề về chất lượng dữ liệu: missing values, duplicates, imbalance

## 🔍 Demo 1.1: Phân Tích Missing Values

Sử dụng `analyze_missing()` để xem chi tiết các cột có giá trị thiếu

```python
# Khởi tạo Clean cho train dataset
cleaner_train = Clean(df_train.copy())

# Phân tích missing values
cleaner_train.analyze_missing()
```

## 🛠️ Demo 1.2: Xử Lý Missing Values

Sử dụng `handle_missing()` để xử lý giá trị thiếu với các phương pháp khác nhau

```python
# Xử lý missing values cho numerical columns với mean
cleaner_train.handle_missing(
    columns=['LotFrontage'], 
    strategy='mean'
)
```

```python
# Xử lý missing values cho categorical columns với mode
cleaner_train.handle_missing(
    columns=['Alley', 'MasVnrType', 'FireplaceQu'], 
    strategy='mode'
)
```

## 🗑️ Demo 1.3: Drop Columns với Missing Values Cao

Xóa các cột có tỉ lệ missing values quá cao (>80%)

```python
# Drop các cột có >80% missing values
cleaner_train.drop_columns_with_missing(threshold=0.8)
```

## 🔄 Demo 1.4: Remove Duplicates

Kiểm tra và loại bỏ các dòng trùng lặp

```python
# Loại bỏ duplicates
cleaner_train.remove_duplicates()
```

## 📊 Demo 1.5: Xem History và Get Data

Xem lịch sử các thao tác đã thực hiện và lấy dữ liệu đã cleaned

```python
# Xem history
cleaner_train.show_history()
```

```python
# Lấy dữ liệu đã cleaned
df_cleaned = cleaner_train.get_cleaned_data()
print(f"\n✅ Cleaned data shape: {df_cleaned.shape}")
df_cleaned.head()
```

# 🔄 PHẦN 2: PREPROCESS MODULE

Module Preprocess xử lý transformation: scaling, encoding, outlier handling

## 📏 Demo 2.1: Scaling Numerical Features

Scale các features số về cùng một range

```python
# Khởi tạo Preprocess với dữ liệu đã cleaned
preprocessor = Preprocess(df_cleaned.copy())

# Scale numerical features với StandardScaler
preprocessor.standardize(
    columns=['LotArea', 'GrLivArea', 'TotalBsmtSF', 'GarageArea'],
    show_comparison=False
)
```

```python
# Thử MinMaxScaler cho một số columns khác
preprocessor.rescale(
    columns=['YearBuilt', 'YearRemodAdd'],
    show_comparison=False
)
```

## 🏷️ Demo 2.2: Encoding Categorical Features

Chuyển đổi categorical features thành dạng số

```python
# Label Encoding cho ordinal features
preprocessor.label_encode(
    columns=['ExterQual', 'ExterCond', 'BsmtQual', 'KitchenQual'],
    show_mapping=False
)
```

```python
# One-Hot Encoding cho nominal features
preprocessor.one_hot_encode(
    columns=['MSZoning', 'Street', 'LotShape'],
    show_result=False
)
```

## 🎯 Demo 2.3: Outlier Detection và Handling

Phát hiện và xử lý outliers

```python
# Remove outliers với IQR method
# Note: Module Preprocess chỉ có remove_outliers (xóa outliers hẳn)
# Ở đây ta sẽ demo phát hiện trước bằng cách xem số lượng outliers
import matplotlib.pyplot as plt

# Phát hiện outliers cho LotArea
col = 'LotArea'
Q1 = df_cleaned[col].quantile(0.25)
Q3 = df_cleaned[col].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers_mask = (df_cleaned[col] < lower) | (df_cleaned[col] > upper)

print(f"📊 Phát hiện outliers trong {col}:")
print(f"   - Q1: {Q1:.2f}, Q3: {Q3:.2f}, IQR: {IQR:.2f}")
print(f"   - Lower bound: {lower:.2f}, Upper bound: {upper:.2f}")
print(f"   - Số outliers: {outliers_mask.sum()} ({outliers_mask.sum()/len(df_cleaned)*100:.2f}%)")
```

```python
# Remove outliers (sẽ xóa các dòng có outliers)
# Lưu ý: thường không nên xóa outliers trong regression vì có thể mất thông tin quan trọng
# Ở đây chỉ demo phương thức, không áp dụng thực tế

# Nếu muốn xóa outliers, uncomment dòng dưới:
# preprocessor.remove_outliers(columns=['LotArea', 'GrLivArea'], method='iqr', show_report=True)

print("⚠️ Đã skip bước remove outliers để giữ nguyên dữ liệu!")
print("💡 Trong thực tế, outliers có thể chứa thông tin quan trọng cho dự đoán giá nhà.")
```

## 📊 Demo 2.4: Xem History và Get Data

Xem lịch sử transformation và lấy dữ liệu đã preprocessed

```python
# Xem history
preprocessor.show_history()
```

```python
# Lấy dữ liệu đã preprocessed
df_preprocessed = preprocessor.get_processed_data()
print(f"\n✅ Preprocessed data shape: {df_preprocessed.shape}")
df_preprocessed.head()
```

# ⭐ PHẦN 3: FEATURE MODULE

Module Feature thực hiện feature selection và engineering

## 🎯 Demo 3.1: Univariate Feature Selection

Chọn features dựa trên statistical tests

```python
# Chuẩn bị dữ liệu cho Feature Selection
# Lấy numerical columns và target
numeric_cols = df_preprocessed.select_dtypes(include=[np.number]).columns.tolist()
if 'SalePrice' in numeric_cols:
    numeric_cols.remove('SalePrice')
if 'Id' in numeric_cols:
    numeric_cols.remove('Id')

# Tạo DataFrame mới kết hợp X và y
X = df_preprocessed[numeric_cols]
y = df_train['SalePrice']  # Target từ train gốc

# Tạo DataFrame hoàn chỉnh với target
df_with_target = X.copy()
df_with_target['SalePrice'] = y.values

# Khởi tạo Feature module với DataFrame có target
feature_selector = Feature(df_with_target, target_col='SalePrice', task='regression')

print(f"✅ Feature selector initialized")
print(f"📊 Data shape: {df_with_target.shape}")
print(f"📊 Features: {len(numeric_cols)}, Target: SalePrice")
```

```python
# Reload module trước khi chạy (để áp dụng fix f_regression)
import importlib
from preprocess.feature import feature
importlib.reload(feature)
from preprocess.feature.feature import Feature

# Xử lý NaN values trước khi feature selection
df_with_target_clean = df_with_target.fillna(df_with_target.median(numeric_only=True))
print(f"✅ Đã xử lý {df_with_target.isna().sum().sum()} NaN values")

# Reinitialize feature_selector với data đã clean
feature_selector = Feature(df_with_target_clean, target_col='SalePrice', task='regression')

# Bây giờ chạy univariate selection với f_regression
feature_selector.univariate_selection(k=20, score_func='f_regression')
```

## 🔄 Demo 3.2: RFE (Recursive Feature Elimination)

Chọn features bằng RFE với Random Forest

```python
# Reset để thử method khác (reset sẽ quay về data gốc)
feature_selector.reset()

# Xử lý NaN sau khi reset
feature_selector.df = feature_selector.df.fillna(feature_selector.df.median(numeric_only=True))
print(f"✅ Đã xử lý NaN sau reset")

# RFE selection với k=15 features
# Method này tự động chọn model phù hợp (LogisticRegression cho classification, LinearRegression cho regression)
feature_selector.rfe_selection(k=15, show_report=True)
```

## 📉 Demo 3.3: PCA (Principal Component Analysis)

Giảm chiều dữ liệu bằng PCA

```python
# Reload module để áp dụng fix PCA
import importlib
from preprocess.feature import feature
importlib.reload(feature)
from preprocess.feature.feature import Feature

# Reset và reinitialize với data clean
feature_selector = Feature(df_with_target_clean, target_col='SalePrice', task='regression')
print(f"✅ Module reloaded và feature_selector reinitialize")

# PCA để giữ 95% variance
feature_selector.pca_extraction(n_components=0.95)
```

## ⭐ Demo 3.4: Feature Importance

Xem importance của features với tree-based model

```python
# Reset và xem feature importance
feature_selector.reset()

# Xử lý NaN sau khi reset
feature_selector.df = feature_selector.df.fillna(feature_selector.df.median(numeric_only=True))
print(f"✅ Đã xử lý NaN sau reset")

# Feature importance với Random Forest
# Method này tự động chọn model phù hợp (RandomForestClassifier cho classification, RandomForestRegressor cho regression)
feature_selector.feature_importance(top_n=15, show_report=True)
```

## 🔗 Demo 3.5: Correlation Analysis

Phân tích correlation giữa các features

```python
# Phân tích correlation (không cần reset nếu chỉ analyze)
# Nhưng nếu muốn fresh start, uncomment dòng dưới:
# feature_selector.reset()
# feature_selector.df = feature_selector.df.fillna(feature_selector.df.median(numeric_only=True))

# Phân tích correlation
feature_selector.correlation_analysis(threshold=0.8)
```

## ➕ Demo 3.6: Add Custom Feature

Thêm feature mới được tính toán từ features hiện có

```python
# Thêm feature mới: TotalSF = TotalBsmtSF + 1stFlrSF + 2ndFlrSF
if 'TotalBsmtSF' in feature_selector.df.columns and '1stFlrSF' in feature_selector.df.columns:
    total_sf = feature_selector.df['TotalBsmtSF'] + feature_selector.df['1stFlrSF']
    if '2ndFlrSF' in feature_selector.df.columns:
        total_sf += feature_selector.df['2ndFlrSF']
    
    feature_selector.add_feature('TotalSF', total_sf)
```

## 📊 Demo 3.7: Xem History và Selected Features

Xem lịch sử và danh sách features đã chọn

```python
# Xem history
feature_selector.show_history()
```

```python
# Xem selected features
selected_features = feature_selector.get_selected_features()
print(f"\n✅ Selected {len(selected_features)} features:")
print(selected_features)
```

```python
# Lấy data với selected features
df_selected = feature_selector.get_data()
print(f"\n✅ Final data shape: {df_selected.shape}")
df_selected.head()
```
