```python
numerical_features = df_train.select_dtypes(include = ['float64', 'int64']).columns
categorical_features =df_train.select_dtypes(include = 'object').columns
print(len(numerical_features),len(categorical_features))
```

```python
eda_train.distribution('SalePrice')
```

```python

target = "SalePrice"
num_features = numerical_features.drop(target)


n_num = len(num_features)
n_cols = 5
n_rows = (n_num // n_cols) + 1

sns.set(style="whitegrid", palette="muted", font_scale=1)

plt.figure(figsize=(20, 4*n_rows))

for i, col in enumerate(num_features, 1):
    plt.subplot(n_rows, n_cols, i)
    sns.scatterplot(
        data=df_train, x=col, y=target,
        alpha=0.6, s=30, color="steelblue", edgecolor=None
    )
    sns.regplot(
        data=df_train, x=col, y=target,
        scatter=False, color="darkred", line_kws={"lw":1.5}
    ) # this adds regression line
    plt.title(f"{col} vs {target}", fontsize=11)

plt.tight_layout()
plt.show()
```

```python
cat_features_no_nulls = [col for col in categorical_features if df_train[col].isnull().sum() == 0]
```

```python
n_cat = len(cat_features_no_nulls)
n_cols = 3
n_rows = (n_cat // n_cols) + 1

plt.figure(figsize=(20, 5*n_rows))

for i, col in enumerate(cat_features_no_nulls, 1):
    plt.subplot(n_rows, n_cols, i)
    sns.boxplot(
        data=df_train, x=col, y=target,
        palette="Set3", showfliers=False
    )
    plt.xticks(fontsize=15)
    plt.title(f"{col} vs {target}", fontsize=11)

plt.tight_layout()
plt.show()
```
